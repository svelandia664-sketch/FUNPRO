```python
def contador():
  calificaciones=[]
  cantidad=int(input("Ingrese la cantidad de calificaciones: "))
  for i in range(cantidad):
    nota=float(input(f"Ingrese las notas {i+1}: "))
    calificaciones.append(nota)
    print(calificaciones)
  c=0
  for a in calificaciones:
    if a>=3:
      c+=1
      print(f"van {c} calificaciones")
  print(f"{c} fueron las calificaciones mayores o iguales 3")
  return cantidad,calificaciones

def promedio(calificaciones,cantidad):
  totnotas=sum(calificaciones)
  promedio=totnotas/cantidad
  print(f"El promedio es: {promedio}")

a,b=contador ()
promedio(b,a)
```
