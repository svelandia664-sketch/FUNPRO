```python
#Ejercicio 1
n=int(input("Ingrese la cantidad de números a analizar "))
cero=0
for i in range(1,n+1):
  num=int(input(f"Ingrese el número {i}: "))
  if num==0: cero=cero+1
print(f"La cantidad de ceros en {n} números es {cero} ")
```

```python
#Ejercicio 2
impar=0
pares=0
n=0

for i in range(10):
  num=int(input("Ingrese un número: "))
  if num%2==0:
    pares=pares+num
    n=n+1
  else: impar=impar+num

print(f"La suma de los números impares es {impar}")
print(f"El promedio de los números pares es {pares/n}")
```

```python
#Ejercicio 3
c=int(input("Ingrese numero de cifras del numero: "))
uno=0

for i in range(c):
  n=int(input(f"Ingrese cifra {i} del numero: "))
  if n==1:
    uno=uno+1
print(f"La cantidad de 1 en su numero es de {uno}.")
```
