```python
import random
import time

# ============================
#   DATOS DEL JUGADOR
# ============================

jugador = {
    "nombre": input("Nombre del héroe: "),
    "nivel": 1,
    "vida": 120,
    "ataque": 25,
    "experiencia": 0
}

compañero = {
    "nombre": "Aliado",
    "nivel": 1,
    "vida": 110,
    "ataque": 20,
    "experiencia": 0
}

# ============================
#   ENEMIGOS
# ============================

enemigos = [
    {"nombre": "Carroñero", "nivel": 1, "vida": 60,  "ataque": 12, "exp": 20},
    {"nombre": "Profesor Molesto", "nivel": 2, "vida": 90, "ataque": 20, "exp": 40},
    {"nombre": "Paladín Caído", "nivel": 3, "vida": 140, "ataque": 30, "exp": 80}
]

# ============================
#   OBJETOS
# ============================

objetos = ["Poción", "Hoja arcana", "Espada vieja"]

# ============================
#   FUNCIONES
# ============================

def enemigo_aleatorio():
    enemigo = random.choice(enemigos).copy()
    print(f"\nAparece {enemigo['nombre']} (Nv {enemigo['nivel']})")
    return enemigo

def atacar(a, b):
    daño = random.randint(a["ataque"] - 5, a["ataque"] + 5)
    b["vida"] -= daño
    print(f"{a['nombre']} hace {daño} dmg.")
    return daño

def drop_objeto():
    if random.random() < 0.25:
        print("Sin botín.")
        return None
    obj = random.choice(objetos)
    print(f"Objeto: {obj}")
    return obj

def ganar_xp(p, cantidad):
    p["experiencia"] += cantidad
    print(f"{p['nombre']} +{cantidad} XP")

# ============================
#   BATALLA
# ============================

def batalla():
    enemigo = enemigo_aleatorio()

    while enemigo["vida"] > 0 and (jugador["vida"] > 0 or compañero["vida"] > 0):

        print("\nTurno del equipo")
        if jugador["vida"] > 0: atacar(jugador, enemigo)
        if enemigo["vida"] > 0 and compañero["vida"] > 0:
            atacar(compañero, enemigo)

        if enemigo["vida"] <= 0:
            break

        print("\nTurno enemigo")
        objetivo = random.choice([p for p in [jugador, compañero] if p["vida"] > 0])
        atacar(enemigo, objetivo)

        print(f"HP enemigo: {max(enemigo['vida'],0)}")
        print(f"HP {jugador['nombre']}: {jugador['vida']} | Aliado: {compañero['vida']}")

    # Resultado
    if enemigo["vida"] <= 0:
        print(f"\n{enemigo['nombre']} derrotado.")
        exp = enemigo["exp"] * enemigo["nivel"]
        print(f"EXP obtenida: {exp}")
        ganar_xp(jugador, exp)
        ganar_xp(compañero, exp)
        drop_objeto()
    else:
        print("\nDerrota...")

# ============================
#   PRINCIPAL
# ============================

print("\nMini RPG — ¡A luchar!\n")

while True:
    x = input("¿Pelear? (s/n): ").lower()
    if x == "s":
        batalla()
    else:
        print("\nFin del juego.")
        print(f"{jugador['nombre']} | HP {jugador['vida']} | XP {jugador['experiencia']}")
        print(f"Aliado | HP {compañero['vida']} | XP {compañero['experiencia']}")
        break
```
